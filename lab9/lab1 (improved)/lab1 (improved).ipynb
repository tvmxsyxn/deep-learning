# PyTorch 101

```python
import torch

# Create a rank 1 tensor from a native Python list
vec_a = torch.tensor([7, 8, 9])
print('Here is vec_a:')
print(vec_a)
print('type(vec_a): ', type(vec_a))
print('rank of vec_a: ', vec_a.dim())
print('vec_a.shape: ', vec_a.shape)

# Access elements using square brackets
print()
print('vec_a[0]: ', vec_a[0])
print('type(vec_a[0]): ', type(vec_a[0]))
print('type(vec_a[0].item()): ', type(vec_a[0].item()))

# Mutate elements using square brackets
vec_a[1] = 99
print()
print('vec_a after mutating:')
print(vec_a)
```

# Creating and Accessing Tensors

```python
# Create a two-dimensional tensor
mat_b = torch.tensor([[2, 4, 6], [8, 10, 12]])
print('Here is mat_b:')
print(mat_b)
print('rank of mat_b:', mat_b.dim())
print('mat_b.shape: ', mat_b.shape)

# Access elements from a multidimensional tensor
print()
print('mat_b[0, 1]:', mat_b[0, 1])
print('mat_b[1, 2]:', mat_b[1, 2])

# Mutate elements of a multidimensional tensor
mat_b[1, 1] = 500
print()
print('mat_b after mutating:')
print(mat_b)
```

# Functions to implement


```python
import torch
from typing import List, Tuple
from torch import Tensor


def build_initial_matrix() -> Tensor:
    mat = torch.zeros(4, 3)
    mat[1, 2] = 42
    mat[2, 0] = 88
    return mat


def apply_tensor_updates(
    target_mat: Tensor, coords: List[Tuple[int, int]], new_vals: List[float]) -> Tensor:
    """
    Update the target tensor based on the provided coordinates and new values.
    """
    for (row, col), val in zip(coords, new_vals):
        target_mat[row, col] = val
    return target_mat


def compute_total_elements(input_tensor: Tensor) -> int:
    """
    Calculate the total number of scalar elements in a given tensor.
    """
    total_count = 1
    for dim_size in input_tensor.shape:
        total_count *= dim_size
    return total_count


def generate_pi_matrix(rows: int, cols: int) -> Tensor:
    """
    Returns a Tensor of shape (rows, cols) filled entirely with the value 3.14.
    """
    result_mat = torch.full((rows, cols), 3.14)
    return result_mat


def extract_tens_multiples(range_start: int, range_end: int) -> Tensor:
    """
    Returns a Tensor of dtype torch.float64 containing all multiples of ten.
    """
    assert range_start <= range_end
    initial_val = ((range_start + 9) // 10) * 10

    if initial_val > range_end:
        return torch.empty(0, dtype=torch.float64)

    result_tensor = torch.arange(initial_val, range_end + 1, step=10, dtype=torch.float64)
    return result_tensor
```

# Testing functions

```python
# Generate a sample tensor
sample_x = build_initial_matrix()
print('Here is the sample tensor:')
print(sample_x)

# Mutate the tensor by setting a few elements
coords_list = [(1, 1), (2, 0), (3, 2)]
vals_list = [14, 25, 36]
apply_tensor_updates(sample_x, coords_list, vals_list)
print('\nAfter mutating:')
print(sample_x)
print('\nCorrect shape: ', sample_x.shape == (4, 3))
print('sample_x[1, 1] correct: ', sample_x[1, 1].item() == 14)
print('sample_x[2, 0] correct: ', sample_x[2, 0].item() == 25)
print('sample_x[3, 2] correct: ', sample_x[3, 2].item() == 36)

# Check the number of elements in the sample tensor
total_num = compute_total_elements(sample_x)
print('\nNumber of elements in sample_x: ', total_num)
print('Correctly counted: ', total_num == 12)

# Test pi tensor
pi_tensor = generate_pi_matrix(5, 6)
print('\npi_tensor is a tensor:', torch.is_tensor(pi_tensor))
print('pi_tensor has correct shape: ', pi_tensor.shape == (5, 6))
print('pi_tensor is filled with pi: ', (pi_tensor == 3.14).all().item() == 1)

# Test multiples of ten
range_start = 12
range_end = 35
tens_result = extract_tens_multiples(range_start, range_end)
print('\nCorrect dtype: ', tens_result.dtype == torch.float64)
print('Correct shape: ', tens_result.shape == (2,))
print('Correct values: ', tens_result.tolist() == [20, 30])

# Empty range test
empty_start = 12
empty_end = 18
empty_result = extract_tens_multiples(empty_start, empty_end)
print('\nCorrect dtype: ', empty_result.dtype == torch.float64)
print('Correct shape: ', empty_result.shape == (0,))
```

# Tensor constructors

```python
# Create a tensor of all zeros
zeros_mat = torch.zeros(3, 4)
print('tensor of zeros:')
print(zeros_mat)

# Create a tensor of all ones
ones_mat = torch.ones(2, 3)
print('\ntensor of ones:')
print(ones_mat)

# Create a 4x4 identity matrix
identity_mat = torch.eye(4)
print('\nidentity matrix:')
print(identity_mat)

# Tensor of random values
random_mat = torch.rand(5, 6)
print('\nrandom tensor:')
print(random_mat)
```

# Datatypes, automatic detection

```python
# Let torch choose the datatype
list_int = torch.tensor([3, 4])      # List of integers
list_float = torch.tensor([3., 4.])  # List of floats
list_mixed = torch.tensor([3., 4])   # Mixed list
print('dtype when torch chooses for us:')
print('List of integers:', list_int.dtype)
print('List of floats:', list_float.dtype)
print('Mixed list:', list_mixed.dtype)
```

# Forced specification

```python
# Force a particular datatype
forced_float = torch.tensor([3, 4], dtype=torch.float32)  # 32-bit float
forced_int32 = torch.tensor([3, 4], dtype=torch.int32)    # 32-bit (signed) integer
forced_int64 = torch.tensor([3, 4], dtype=torch.int64)    # 64-bit (signed) integer
print('\ndtype when we force a datatype:')
print('32-bit float: ', forced_float.dtype)
print('32-bit integer: ', forced_int32.dtype)
print('64-bit integer: ', forced_int64.dtype)

# Other creation ops also take a dtype argument
default_ones = torch.ones(2, 3)                      # Let torch choose for us
int16_ones = torch.ones(2, 3, dtype=torch.int16)     # 16-bit (signed) integer
uint8_ones = torch.ones(2, 3, dtype=torch.uint8)     # 8-bit (unsigned) integer
print('\ntorch.ones with different dtypes')
print('default dtype:', default_ones.dtype)
print('16-bit integer:', int16_ones.dtype)
print('8-bit unsigned integer:', uint8_ones.dtype)
```

# Type casting

```python
base_eye = torch.eye(4, dtype=torch.int64)
cast_float32 = base_eye.float()               # Cast to 32-bit float
cast_float64 = base_eye.double()              # Cast to 64-bit float
alt_cast_f32 = base_eye.to(torch.float32)     # Alternate way to cast to 32-bit float
alt_cast_f64 = base_eye.to(torch.float64)     # Alternate way to cast to 64-bit float
print('base_eye:', base_eye.dtype)
print('cast_float32:', cast_float32.dtype)
print('cast_float64:', cast_float64.dtype)
print('alt_cast_f32:', alt_cast_f32.dtype)
print('alt_cast_f64:', alt_cast_f64.dtype)
```

# Preserving Dtype

```python
source_tensor = torch.eye(4, dtype=torch.float64)   # Shape (4, 4), dtype torch.float64
zeros_like = torch.zeros_like(source_tensor)        # Shape (4, 4), dtype torch.float64
new_zeros = source_tensor.new_zeros(5, 6)           # Shape (5, 6), dtype torch.float64
ones_to = torch.ones(7, 8).to(source_tensor)        # Shape (7, 8), dtype torch.float64
print('source_tensor shape is %r, dtype is %r' % (source_tensor.shape, source_tensor.dtype))
print('zeros_like shape is %r, dtype is %r' % (zeros_like.shape, zeros_like.dtype))
print('new_zeros shape is %r, dtype is %r' % (new_zeros.shape, new_zeros.dtype))
print('ones_to shape is %r, dtype is %r' % (ones_to.shape, ones_to.dtype))
```
